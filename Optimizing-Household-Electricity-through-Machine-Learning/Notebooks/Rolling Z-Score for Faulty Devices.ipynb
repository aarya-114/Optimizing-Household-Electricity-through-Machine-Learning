```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
df1 = pd.read_csv(r'/Users/aaryapatil/Desktop/esssssay/Optimizing-Household-Electricity-through-Machine-Learning/Datasets/electricity_appliance_wise_data.csv')
df1['Date'] = pd.to_datetime(df1['Date'])
df1 = df1[df1['Date'].dt.year == 2021]
df1.reset_index(drop=True, inplace=True)
df1.head(2)
```

```python
def zscore(x, window):
    r = x.rolling(window=window)
    m = r.mean().shift(1)
    s = r.std(ddof=0).shift(1)
    z = (x-m)/s
    return z
```

```python
df1['kap_zscore'] = zscore(df1['Kitchen Appliances'], 30)
df1['fridge_zscore'] = zscore(df1['Fridge'], 30)
df1['ac_zscore'] = zscore(df1['AC'], 30)
df1['oap_zscore'] = zscore(df1['Other Appliances'], 30)
df1['wm_zscore'] = zscore(df1['Washing Machine'], 3)
```
